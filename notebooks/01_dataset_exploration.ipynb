# NoteGuard AI - Notebook 01: Dataset Exploration & Integrity Analysis

### Objectives:
1. Inspect raw Indian currency note dataset hierarchy.
2. Verify image dimensions, file formats, and channel configurations.
3. Analyze class balance between Genuine and Fake samples across denominations (₹100, ₹200, ₹500).
4. Detect potential data leakage vectors and compute MD5 deduplication hashes.

```python
import os
import glob
from pathlib import Path
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

RAW_DIR = Path('../dataset/raw')
print(f'Raw dataset directory: {RAW_DIR.resolve()}')
```

### 1. File Discovery & Extension Verification

```python
samples = []
for p in RAW_DIR.glob('**/*.*'):
    if p.is_file() and p.suffix.lower() in ['.jpg', '.jpeg', '.png', '.webp']:
        parts = [part.lower() for part in p.parts]
        auth = 'genuine' if 'genuine' in parts else 'fake'
        denom = [d for d in ['100', '200', '500', '2000'] if d in parts]
        samples.append({
            'file': str(p),
            'denomination': denom[0] if denom else 'unknown',
            'authenticity': auth,
            'size_kb': round(p.stat().st_size / 1024, 2)
        })

df = pd.DataFrame(samples)
print(f'Total raw samples discovered: {len(df)}')
df.head()
```

### 2. Class Distribution Visualizations

```python
if not df.empty:
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    df['authenticity'].value_counts().plot(kind='bar', color=['#10b981', '#ef4444'])
    plt.title('Distribution by Authenticity')
    plt.ylabel('Count')

    plt.subplot(1, 2, 2)
    df['denomination'].value_counts().plot(kind='bar', color=['#3b82f6', '#8b5cf6', '#ec4899'])
    plt.title('Distribution by Denomination')
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()
```
